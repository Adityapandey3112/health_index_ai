```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
dataset = pd.read_csv('food1.csv')
#Data Preprocessing
dataset = dataset[dataset["Dish Eaten"] != "Nothing"]
dataset = dataset.replace(to_replace = 'n/a', value = np.nan)
dataset = dataset.ffill()

#FSSAI Daily Reference Values Per Diner Meal
fssai_daily_ref = {
    'Total Calories (kcal)': 600.0,  # FSSAI 2,000 kcal/day reference
    'Protein Content(in g)': 18.0,  # 54 g/day adult reference
    'Fibre Content(in g)': 10.0,  # 30 g/day dietary fibre reference
    'Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)': 3.2, # 10g/day reference
}
#Data Visualisation
a = dataset['Total Calories (kcal)'].values
b = dataset['Protein Content(in g)'].values
c = dataset['Fibre Content(in g)'].values
d = dataset['Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)'].values
x = np.arange(1, len(a)+1, 1)
fig = plt.figure(figsize =(14, 14))
plt.suptitle('Nutritional Content per Meal Chart', fontsize=20)
sub1 = plt.subplot(2, 2, 1)
sub2 = plt.subplot(2, 2, 2)
sub3 = plt.subplot(2, 2, 3)
sub4 = plt.subplot(2, 2, 4)
sub1.plot(x, a, 'sb')
sub1.set_title('Calories Chart')
sub1.set_xlabel('Order Number')
sub1.set_ylabel('Calories this Meal')
sub2.plot(x, b, 'or')
sub2.set_title('Protein Chart')
sub2.set_xlabel('Order Number')
sub2.set_ylabel('Protein this Meal')
sub3.plot(x, c,  'Dm')
sub3.set_title('Fibre Chart')
sub3.set_xlabel('Order Number')
sub3.set_ylabel('Fibres this Meal')
sub4.plot(x, d, 'g^')
sub4.set_title('Vitamins, Minerals and Antioxidants Chart')
sub4.set_xlabel('Order Number')
sub4.set_ylabel('Vitamins, Minerals and Antioxidants this Meal')
sub1.axhline(
    fssai_daily_ref['Total Calories (kcal)'],
    color='red',
    linestyle='-',
    linewidth=1.8,
    label='FSSAI Ref (600.0)',
)
sub2.axhline(
    fssai_daily_ref['Protein Content(in g)'],
    color='green',
    linestyle='-',
    linewidth=1.8,
    label='FSSAI Ref (18.0)',
)
sub3.axhline(
    fssai_daily_ref['Fibre Content(in g)'],
    color='pink',
    linestyle='-',
    linewidth=1.8,
    label='FSSAI Ref (10.0)',
)
sub4.axhline(
    fssai_daily_ref[
        'Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)'
    ],
    color='orange',
    linestyle='-',
    linewidth=1.8,
    label='FSSAI Ref (3.2)',
)
sub1.legend()
sub2.legend()
sub3.legend()
sub4.legend()
plt.show()
print('\n\n\n\n')

print('The average amount of calories in each meal is:', np.mean(a))
print('The average amount of protein in each meal is:', np.mean(b))
print('The average amount of fibre in each meal is:', np.mean(c))
print('The average amount of vitamins, minerals and antioxidants in each meal is:', np.mean(d))
print('\n\n')
print('The daily fssai amount for calories in each meal is:', fssai_daily_ref['Total Calories (kcal)'])
print('The daily fssai amount for protein in each meal is:', fssai_daily_ref['Protein Content(in g)'])
print('The daily fssai amount for fibre in each meal is:', fssai_daily_ref['Fibre Content(in g)'])
print('The daily fssai amount for vitamins, minerals and antioxidants in each meal is:', fssai_daily_ref['Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)'])

```
