```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
dataset = pd.read_csv('food1.csv')
#Data Preprocessing
dataset = dataset[dataset["Dish Eaten"] != "Nothing"]
dataset = dataset.replace(to_replace = 'n/a', value = np.nan)
dataset = dataset.ffill()
x = dataset.iloc[:, 5:11]
y = dataset.iloc[:, -1].map({'Healthy': 1, 'Unhealthy': 0})
#Data Visualisation
a = dataset['Total Calories (kcal)'].values
b = dataset['Protein Content(in g)'].values
c = dataset['Fibre Content(in g)'].values
d = dataset['Added Sugar(in g)'].values
e = dataset['Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)'].values
fig = plt.figure(figsize =(20, 20))
plt.suptitle('Food Health Chart and Correlations', fontsize=20)
sub1 = plt.subplot(3, 2, 1)
sub2 = plt.subplot(3, 2, 2)
sub3 = plt.subplot(3, 2, 3)
sub4 = plt.subplot(3, 2, 4)
sub5 = plt.subplot(3, 2, 5)
sub1.plot(a, b, 'sb')
sub1.set_title('Calories vs Protein Content')
sub1.set_xlabel('Calories')
sub1.set_ylabel('Protein Content')
sub2.plot(c, e, 'or')
sub2.set_title('Fibre Content vs Vitamins, Minerals and Antioxidants')
sub2.set_xlabel('Fibre Content')
sub2.set_ylabel('Vitamins, Minerals and Antioxidants')
sub3.plot(c, a,  'Dm')
sub3.set_title('Total Fibre vs Total Calories')
sub3.set_xlabel('Total Fibre')
sub3.set_ylabel('Total Calories')
sub4.plot(d, b, 'g^')
sub4.set_title('Key Vitamins, Minerals and Antioxidants vs Protein Content')
sub4.set_xlabel('Key Vitamins, Minerals and Antioxidants')
sub4.set_ylabel('Protein Content')
sub5.plot(e, b, '+r')
sub5.set_title('Vitamin, Mineral, Antioxidant Quantity vs Protein Content')
sub5.set_xlabel('Vitamin, Mineral, Antioxidant Quantity')
sub5.set_ylabel('Protein Content')
plt.show()
print('\n\n\n\n')
import seaborn as sns
corr = dataset.iloc[:, 5:11].corr()
sns.heatmap(corr, annot = True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Matrix")
plt.show()
print('\n\n\n\n')

#Machine Learning Part
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(x)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 21)
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier(n_estimators = 10, criterion = 'gini', random_state = 21)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
acc = accuracy_score(y_test, y_pred)
print('\n\n\n\n')
print(f"Accuracy is {round(acc*100, 2)}")
print(f"Misclassification is {round((1-acc)*100, 2)}")
cm = confusion_matrix(y_test, y_pred)
import matplotlib.pyplot as plt
disp = ConfusionMatrixDisplay(confusion_matrix = cm, display_labels = y_test.unique())
disp.plot(cmap = plt.cm.Blues)
plt.title("Food Health Index")
plt.show()

print('\n\n\n\n')
print('Classififation Report')
print(classification_report(y_test, y_pred))
print('\n\n\n\n')
import statsmodels.api as sm
X_train_sm = sm.add_constant(X_train)
l_r = sm.OLS(y_train, X_train_sm).fit()
print(l_r.summary())
```
