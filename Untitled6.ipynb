```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Load raw diner meal dataset
dataset = pd.read_csv('food1.csv')

#Data
# Remove records where no meal was consumed (retaining N = 75 valid meal orders)
dataset = dataset[dataset["Dish Eaten"] != "Nothing"]
# Standardize string 'n/a' markers into NaN values
dataset = dataset.replace(to_replace = 'n/a', value = np.nan)
# Forward-fill missing entries using the preceding valid meal observation
dataset = dataset.ffill()
dataset = dataset.head(75)

#FSSAI Daily Reference Values Per Diner Meal
fssai_daily_ref = {
    'Total Calories (kcal)': 600.0,  # FSSAI 2,000 kcal/day reference
    'Protein Content(in g)': 18.0,  # 54 g/day adult reference
    'Fibre Content(in g)': 10.0,  # 30 g/day dietary fibre reference
    'Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)': 3.2, # 10g/day reference
}

#Data Visualisation
a = dataset['Total Calories (kcal)'].values
b = dataset['Protein Content(in g)'].values
c = dataset['Fibre Content(in g)'].values
d = dataset['Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)'].values
x = np.arange(1, len(a)+1, 1)
fig = plt.figure(figsize =(14, 10))
plt.suptitle('Nutritional Content per Meal Chart (N = 75 Orders)', fontsize=16)
sub1 = plt.subplot(2, 2, 1)
sub2 = plt.subplot(2, 2, 2)
sub3 = plt.subplot(2, 2, 3)
sub4 = plt.subplot(2, 2, 4)
sub1.plot(x, a,  marker='s', color='#1E40AF', label='Meal Calories')
sub1.set_ylabel('Calories this Meal')
sub2.plot(x, b,  marker='o', color='#0F766E', label='Meal Protein')
sub2.set_title('Protein Chart')
sub2.set_ylabel('Protein this Meal')
sub3.plot(x, c,   marker='D', color='#B45309', label='Meal Fibre')
sub3.set_title('Fibre Chart')
sub3.set_ylabel('Fibres this Meal')
sub4.plot(x, d,  marker='^', color='#6D28D9', label='Meal Micronutrients')
sub4.set_title('Vitamins, Minerals and Antioxidants Chart')
sub4.set_ylabel('Vitamins, Minerals and Antioxidants this Meal')
sub1.axhline(fssai_daily_ref['Total Calories (kcal)'], color='#B91C1C', linestyle='--', label='FSSAI Ref (600.0)')
sub2.axhline(fssai_daily_ref['Protein Content(in g)'], color='#B91C1C', linestyle='--', label='FSSAI Ref (18.0)')
sub3.axhline(fssai_daily_ref['Fibre Content(in g)'], color='#B91C1C', linestyle='--', label='FSSAI Ref (10.0)')
sub4.axhline(fssai_daily_ref['Key Vitamins, Minerals & Antioxidants(Total Quantity in grams)'], color='#B91C1C', linestyle='-', label='FSSAI Ref (3.2)')
for ax in [sub1, sub2, sub3, sub4]:
  ax.set_xlabel('Order Number (1 - 75)')
ax.legend()
sub1.legend()
sub2.legend()
sub3.legend()
sub4.legend()
plt.show()
print('\n\n\n\n')

# Core mean comparison across 75 meal orders
print('The average amount of calories in each meal is:', round(np.mean(a), 2), 'kcal')
print('The average amount of protein in each meal is:', round(np.mean(b), 2), 'g')
print('The average amount of fibre in each meal is:', round(np.mean(c), 2), 'g')
print('The average amount of vitamins, minerals and antioxidants in each meal is:', round(np.mean(d), 2), 'g')
# Extended analysis: FSSAI target compliance rate (%)
summary_stats = pd.DataFrame({
    'Metric': ['Calories (kcal)', 'Protein (g)', 'Fibre (g)', 'Vitamins & Minerals (g)'],
    'Mean': [np.mean(a), np.mean(b), np.mean(c), np.mean(d)],
    'Std_Dev': [np.std(a), np.std(b), np.std(c), np.std(d)],
    'FSSAI_Ref': [600.0, 18.0, 10.0, 3.2],
    'Meals_Meeting_Target_%': [
        np.mean(a >= 600.0) * 100,
        np.mean(b >= 18.0) * 100,
        np.mean(c >= 10.0) * 100,
        np.mean(d >= 3.2) * 100  ]
    })
print(summary_stats.round(2))
```
