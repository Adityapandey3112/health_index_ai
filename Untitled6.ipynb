```python
from re import sub
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
dataset = pd.read_csv('food1.csv')
x = dataset.iloc[:, 2:8]
y = dataset.iloc[:, -1].map({'Healthy': 1, 'Unhealthy': 0})
dataset = dataset.replace(to_replace = 'n/a', value = np.nan)
dataset = dataset.ffill()
a = dataset['Total Calories (kcal)'].values
b = dataset['Protein Content'].values
c = dataset['Fibre Content'].values
d = dataset['Added Sugar'].values
e = dataset['Key Vitamins, Minerals & Antioxidants'].values
fig = plt.figure(figsize =(20, 20))
sub1 = plt.subplot(3, 2, 1)
sub2 = plt.subplot(3, 2, 2)
sub3 = plt.subplot(3, 2, 3)
sub4 = plt.subplot(3, 2, 4)
sub5 = plt.subplot(3, 2, 5)
sub1.plot(a, b, 'sb-')
sub1.set_title('Calories vs Protein Content')
sub1.set_xlabel('Calories')
sub1.set_ylabel('Protein Content')
sub2.plot(b, y, 'or-')
sub2.set_title('Protein Content vs Health Index')
sub2.set_xlabel('Protein Content')
sub2.set_ylabel('Health Index')
sub3.plot(c, y,  'Dm-')
sub3.set_title('Total Fibre vs Health Index')
sub3.set_xlabel('Total Fibre')
sub3.set_ylabel('Health Index')
sub4.plot(d, y, 'g^-')
sub4.set_title('Added Sugar vs Health Index')
sub4.set_xlabel('Added Sugar')
sub4.set_ylabel('Health Index')
sub5.plot(e, y, 'Dr-')
sub5.set_title('Vitamin, Mineral, Antioxidant Quantity vs Health Index')
sub5.set_xlabel('Vitamin, Mineral, Antioxidant Quantity')
sub5.set_ylabel('Health Index')
plt.show()
import seaborn as sns
corr = dataset.iloc[:, 2:8].corr()
sns.heatmap(corr, annot = True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Matrix")
plt.show()
plot_df = dataset.copy()
plot_df['Health_Status'] = y.map({1: 'Healthy', 0: 'Unhealthy'})

sns.pairplot(plot_df, hue='Health_Status', vars=['Total Calories (kcal)', 'Protein Content', 'Fibre Content', 'Added Sugar', 'Key Vitamins, Minerals & Antioxidants'], palette='Set1')
plt.show()
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(x)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 21)
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier(n_estimators = 5, criterion = 'gini', random_state = 21)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy is {round(acc*100, 2)}")
print(f"Misclassification is {round((1-acc)*100, 2)}")
cm = confusion_matrix(y_test, y_pred)
import matplotlib.pyplot as plt
disp = ConfusionMatrixDisplay(confusion_matrix = cm, display_labels = y_test.unique())
disp.plot(cmap = plt.cm.Blues)
plt.title("Food Health Index")
plt.show()
```
